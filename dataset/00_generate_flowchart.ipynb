# Generate Flowchart untuk Jurnal

Flowchart metodologi penelitian lengkap dengan sub-proses

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Ellipse
import numpy as np

fig, ax = plt.subplots(figsize=(16, 24))
ax.set_xlim(0, 16)
ax.set_ylim(0, 24)
ax.axis('off')

def draw_box(x, y, w, h, text, fontsize=10, color='white', edgecolor='black'):
    box = FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.05",
                         facecolor=color, edgecolor=edgecolor, linewidth=2)
    ax.add_patch(box)
    ax.text(x + w/2, y + h/2, text, ha='center', va='center', fontsize=fontsize, 
            fontweight='bold', wrap=True)

def draw_ellipse(x, y, w, h, text, fontsize=11, color='#e3f2fd'):
    ellipse = Ellipse((x, y), w, h, facecolor=color, edgecolor='black', linewidth=2)
    ax.add_patch(ellipse)
    ax.text(x, y, text, ha='center', va='center', fontsize=fontsize, fontweight='bold')

def draw_arrow(x1, y1, x2, y2):
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', color='black', lw=2))

# Title
ax.text(8, 23.5, 'FLOWCHART METODOLOGI PENELITIAN', ha='center', fontsize=16, fontweight='bold')
ax.text(8, 23, 'Ekstraksi Kata Kunci dengan N-Gram dan IndoBERT untuk\nRekomendasi Wisata Bogor', 
        ha='center', fontsize=11, style='italic')

# ============ MULAI ============
draw_ellipse(8, 22, 2, 0.7, 'MULAI', color='#c8e6c9')
draw_arrow(8, 21.6, 8, 21)

# ============ SCRAPPING ============
draw_box(5.5, 19.5, 5, 1.4, 'SCRAPPING DATA\n• Selenium WebDriver\n• Google Maps API\n• TripAdvisor', fontsize=9, color='#fff9c4')
draw_arrow(8, 19.5, 8, 18.8)

# ============ DATA WISATA BOGOR ============
draw_box(5.5, 17.5, 5, 1.2, 'DATA WISATA BOGOR\n(CSV Dataset)', fontsize=10, color='#e8f4f8')
draw_arrow(8, 17.5, 8, 16.8)

# ============ PREPROCESSING ============
draw_box(4.5, 14.5, 7, 2.2, 'PREPROCESSING\n• Case Folding (lowercase)\n• Remove URL\n• Remove Numbers\n• Remove Special Characters\n• Remove Extra Whitespace', fontsize=9, color='#fff3cd')
draw_arrow(8, 14.5, 8, 13.8)

# Split label
ax.text(8, 13.9, '▼', ha='center', fontsize=14)

# ============ TWO PATHS ============
# Left arrow
draw_arrow(8, 13.5, 4.5, 12.5)
# Right arrow  
draw_arrow(8, 13.5, 11.5, 12.5)

# ===== PATH 1: N-GRAM (Left) =====
ax.text(4.5, 13, 'PATH 1: N-GRAM', ha='center', fontsize=10, fontweight='bold', color='#2e7d32')

draw_box(2, 10.5, 5, 1.8, 'EKSTRAKSI N-GRAM\n• Unigram (n=1)\n• Bigram (n=2)\n• Trigram (n=3)', fontsize=9, color='#c8e6c9')
draw_arrow(4.5, 10.5, 4.5, 9.8)

draw_box(2, 8, 5, 1.7, 'TF-IDF VECTORIZATION\n• Max Features: 5000\n• N-gram Range: (1,3)\n• Sublinear TF', fontsize=9, color='#c8e6c9')
draw_arrow(4.5, 8, 4.5, 7.3)

draw_box(2, 5.5, 5, 1.7, 'COSINE SIMILARITY\n(Lexical Similarity)\nBerdasarkan Kata Kunci', fontsize=9, color='#c8e6c9')
ax.text(4.5, 5, '50%', ha='center', fontsize=12, fontweight='bold', color='#2e7d32',
        bbox=dict(boxstyle='round', facecolor='white', edgecolor='#2e7d32'))

# ===== PATH 2: INDOBERT (Right) =====
ax.text(11.5, 13, 'PATH 2: INDOBERT', ha='center', fontsize=10, fontweight='bold', color='#1565c0')

draw_box(9, 10.5, 5, 1.8, 'TOKENIZER INDOBERT\n• indobert-base-p1\n• Vocab Size: 50000\n• Max Length: 128', fontsize=9, color='#bbdefb')
draw_arrow(11.5, 10.5, 11.5, 9.8)

draw_box(9, 8, 5, 1.7, 'TRAINING dengan SimCSE\n• MultipleNegatives\n   RankingLoss\n• Contrastive Learning', fontsize=9, color='#bbdefb')
draw_arrow(11.5, 8, 11.5, 7.3)

draw_box(9, 5.5, 5, 1.7, 'GENERATE EMBEDDING\n• Mean Pooling\n• Dimension: 768\n• Save .npy', fontsize=9, color='#bbdefb')
draw_arrow(11.5, 5.5, 11.5, 4.8)

draw_box(9, 3.2, 5, 1.5, 'COSINE SIMILARITY\n(Semantic Similarity)\nBerdasarkan Makna', fontsize=9, color='#bbdefb')
ax.text(11.5, 2.7, '50%', ha='center', fontsize=12, fontweight='bold', color='#1565c0',
        bbox=dict(boxstyle='round', facecolor='white', edgecolor='#1565c0'))

# ============ MERGE ============
# Arrows merging
draw_arrow(4.5, 5, 8, 2.3)
draw_arrow(11.5, 2.7, 8, 2.3)

# Combined
draw_box(5, 1, 6, 1.2, 'KOMBINASI SIMILARITY\n50% N-gram + 50% IndoBERT', fontsize=10, color='#f3e5f5')
draw_arrow(8, 1, 8, 0.3)

# Move rest down
# We need more space, let me adjust

plt.tight_layout()
plt.savefig('./data/flowchart_part1.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("\n✅ Part 1 saved!")
```
